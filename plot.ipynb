```python
import json
import pandas as pd
import numpy as np
from functools import reduce
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors
# import seaborn as sns
```

# ResNet18

```python
file_path = 'stats/val/performance_resnet18_stats.json'
with open(file_path, 'r') as f:
    performance_data = json.load(f)
    
epochs_resnet18 = performance_data.get("all_epoch_scores", [])
epochs_resnet18 = pd.DataFrame(epochs_resnet18)
print(epochs_resnet18.head())

best_epoch_resnet18 = performance_data.get('best_epoch_stats', {})
columns = list(best_epoch_resnet18.keys())
values = list(best_epoch_resnet18.values())
best_epoch_resnet18 = np.array([values])
print("Columns:", columns)
print("Best Epoch Array:", best_epoch_resnet18)

file_path = 'stats/test/performance_resnet18_stats.json'
with open(file_path, 'r') as f:
    performance_data = json.load(f)
    
test_resnet18 = performance_data.get('test_stats', [])
columns = list(test_resnet18.keys())
values = list(test_resnet18.values())
test_resnet18 = np.array([values])
print("Columns:", columns)
print("Test Result Array:", test_resnet18)

cmatrix_resnet18 = performance_data.get('confusion_matrix', [])
cmatrix_resnet18 = np.array(cmatrix_resnet18)
print("Confusion Matrix:")
print(cmatrix_resnet18)
```

# Comparison

```python
f1_resnet18 = epochs_resnet18[["epoch", "f1"]].rename(columns={"f1": "ResNet18"})
f1_resnet50 = epochs_resnet50[["epoch", "f1"]].rename(columns={"f1": "ResNet50"})
f1_efficientnetb0 = epochs_efficientnetb0[["epoch", "f1"]].rename(columns={"f1": "EfficientNetB0"})
f1_efficientnetb2 = epochs_efficientnetb2[["epoch", "f1"]].rename(columns={"f1": "EfficientNetB2"})

f1_epochs = reduce(
    lambda left, right: pd.merge(left, right, on="epoch"),
    [f1_resnet18, f1_resnet50, f1_efficientnetb0, f1_efficientnetb2]
)

f1_epochs.head()
```

```python
cmap = cm.get_cmap('RdBu', len(f1_epochs.columns) - 1)
colors = [mcolors.to_hex(cmap(i)) for i in range(len(f1_epochs.columns) - 1)]

plt.figure(figsize=(10, 4))
for idx, model in enumerate(f1_epochs.columns[1:]):
    plt.plot(f1_epochs["epoch"], f1_epochs[model], label=model, color=colors[idx])
plt.title("F1 Score Across Epochs for Models")
plt.xlabel("Epoch")
plt.ylabel("F1 Score")
plt.xticks()
plt.legend()
plt.show()
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1-Score"]
model1_values = best_epoch_resnet18[0, 1:]  
model2_values = best_epoch_resnet50[0, 1:]  

bar_width = 0.35
x = np.arange(len(metrics))

plt.figure(figsize=(8, 5))
plt.bar(x - bar_width/2, model1_values, bar_width, label="ResNet18", color="#003366")
plt.bar(x + bar_width/2, model2_values, bar_width, label="ResNet50", color="#B22222")

for i, (m1, m2) in enumerate(zip(model1_values, model2_values)):
    plt.text(x[i] - bar_width/2, m1 + 0.01, f"{m1:.2f}", ha="center", va="bottom", color="black")
    plt.text(x[i] + bar_width/2, m2 + 0.01, f"{m2:.2f}", ha="center", va="bottom", color="black")

plt.xticks(x, metrics)
plt.title("ResNet Validation Performance")
plt.xlabel("Metrics")
plt.ylabel("Value")
plt.legend()
plt.ylim(0, 1.2)
plt.tight_layout()
plt.show()
```

```python
# Function to plot a heatmap
def plot_heatmap(matrix, cmap, title, xticklabels, yticklabels):
    plt.imshow(matrix, interpolation="nearest", cmap=cmap)
    plt.title(title)
    plt.colorbar()
    plt.xticks(ticks=np.arange(len(xticklabels)), labels=xticklabels, fontsize=10)
    plt.yticks(ticks=np.arange(len(yticklabels)), labels=yticklabels, fontsize=10)
    plt.xlabel("Prediction")
    plt.ylabel("Ground Truth")
    
    # Annotate the cells with values
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            plt.text(j, i, f"{matrix[i, j]}", ha="center", va="center", color="black", fontsize=12, fontweight="bold")

# Plot the heatmaps
plt.figure(figsize=(9, 4))

plt.subplot(1, 2, 1)
plot_heatmap(
    cmatrix_resnet18, cmap="Blues",
    title="ResNet18 Confusion Matrix",
    xticklabels=["Different", "Same"],
    yticklabels=["Different", "Same"]
)

plt.subplot(1, 2, 2)
plot_heatmap(
    cmatrix_resnet50, cmap="Reds",
    title="ResNet50 Confusion Matrix",
    xticklabels=["Different", "Same"],
    yticklabels=["Different", "Same"]
)

plt.tight_layout()
plt.show()
```
